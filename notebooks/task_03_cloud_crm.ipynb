## 1. Постановка задачи

Компания «ТехноПром» рассматривает внедрение облачной CRM для отдела продаж. Нужно:

1. рассчитать экономию времени на рутинных операциях и её денежную оценку;
2. оценить дополнительный доход от роста продаж;
3. определить стоимость лицензий и потери на период адаптации;
4. построить денежные потоки за три года с учётом налога;
5. рассчитать NPV, простой срок окупаемости и ROI;

## 2. Исходные данные

| Параметр | Значение |
|---|---:|
| Количество менеджеров | 15 чел. |
| Средняя зарплата менеджера | 120 000 руб./мес. |
| Экономия времени на рутинных операциях после CRM | 20% от времени на рутину |
| Доля рабочего времени на рутинные операции до CRM | 30% |
| Прогнозируемый рост продаж | 8% в год |
| Текущий годовой оборот отдела продаж | 72 000 000 руб. |
| Стоимость CRM на пользователя | 24 000 руб./год |
| Период адаптации | 1 месяц |
| Горизонт оценки | 3 года |
| Ставка дисконтирования | 12% в год |
| Налог на прибыль | 20% |

### Принятые допущения

- Экономия рабочего времени оценивается в денежном выражении, но сама по себе не означает сокращение фонда оплаты труда: высвободившееся время должно быть направлено на полезные задачи.
- Дополнительный оборот 5,76 млн руб. рассматривается как прямой эффект по условию задачи. Маржинальность продаж не дана, поэтому расчёт не учитывает себестоимость дополнительной выручки. В реальном проекте вместо выручки следует использовать дополнительную маржинальную прибыль.
- Лицензии оплачиваются ежегодно. Потери на адаптацию учитываются единовременно в момент запуска (год 0) как один месячный фонд оплаты труда.
- Для годовых денежных потоков налог рассчитывается с эффекта за вычетом годовой стоимости лицензий. Налоговый эффект затрат на адаптацию в году 0 не учитывается.

```python
# Исходные параметры проекта

managers = 15
monthly_salary = 120_000
routine_share_before = 0.30
routine_time_saving = 0.20
sales_growth = 0.08
annual_turnover = 72_000_000
crm_cost_per_user_year = 24_000
adaptation_months = 1
evaluation_years = 3
discount_rate = 0.12
profit_tax_rate = 0.20

print(f"Количество менеджеров: {managers}")
print(f"Фонд оплаты труда в месяц: {managers * monthly_salary:,.0f} руб.")
print(f"Годовой оборот отдела продаж: {annual_turnover:,.0f} руб.")
print(f"Горизонт оценки: {evaluation_years} года")
```

## 3. Прямой экономический эффект

### 3.1. Экономия времени на рутинных операциях

Сначала рассчитывается годовой фонд оплаты труда (ФОТ). Доля рутинного времени после внедрения CRM уменьшается на 20% относительно исходной доли 30%: 30% × (1 − 20%) = 24%. Таким образом, высвобождается 6 процентных пунктов рабочего времени.

```python
annual_payroll = managers * monthly_salary * 12
routine_share_after = routine_share_before * (1 - routine_time_saving)
freed_time_share = routine_share_before - routine_share_after
annual_labor_time_saving_value = annual_payroll * freed_time_share

print(f"ФОТ за год: {annual_payroll:,.0f} руб.")
print(f"Доля рутины до CRM: {routine_share_before:.0%}")
print(f"Доля рутины после CRM: {routine_share_after:.0%}")
print(f"Высвобождаемое рабочее время: {freed_time_share:.0%}")
print(f"Денежная оценка экономии времени: {annual_labor_time_saving_value:,.0f} руб./год")
```

### 3.2. Дополнительный доход от роста продаж

Дополнительный оборот рассчитывается как 8% от текущего годового оборота. Поскольку рост указан как ежегодный прогноз, в базовом сценарии для каждого из трёх лет используется одинаковый эффект относительно текущего оборота — так же, как в решении из методички.

```python
additional_sales_revenue = annual_turnover * sales_growth
annual_gross_effect = annual_labor_time_saving_value + additional_sales_revenue

print(f"Дополнительный оборот: {additional_sales_revenue:,.0f} руб./год")
print(f"Суммарный прямой эффект до затрат и налогов: {annual_gross_effect:,.0f} руб./год")
```

## 4. Затраты на внедрение

```python
annual_crm_licenses = managers * crm_cost_per_user_year
total_license_cost = annual_crm_licenses * evaluation_years
monthly_payroll = managers * monthly_salary
adaptation_loss = monthly_payroll * adaptation_months

print(f"Стоимость лицензий в год: {annual_crm_licenses:,.0f} руб.")
print(f"Стоимость лицензий за {evaluation_years} года: {total_license_cost:,.0f} руб.")
print(f"Потери производительности на адаптации (год 0): {adaptation_loss:,.0f} руб.")
print(f"Общие затраты за период без учёта налогов: {total_license_cost + adaptation_loss:,.0f} руб.")
```

## 5. Денежные потоки и прирост чистой прибыли

Для каждого года 1–3 из суммарного прямого эффекта вычитается годовая лицензия. С полученной прибыли рассчитывается налог 20%. Чистый годовой денежный поток равен прибыли после налога. В году 0 учитывается разовая потеря на адаптацию.

```python
annual_profit_before_tax = annual_gross_effect - annual_crm_licenses
annual_tax = annual_profit_before_tax * profit_tax_rate
annual_net_cash_flow = annual_profit_before_tax - annual_tax

cash_flows = [-adaptation_loss] + [annual_net_cash_flow] * evaluation_years

cash_flow_df = pd.DataFrame({
    "Год": list(range(evaluation_years + 1)),
    "Прямой эффект, руб.": [0] + [annual_gross_effect] * evaluation_years,
    "Затраты на лицензии, руб.": [0] + [annual_crm_licenses] * evaluation_years,
    "Прибыль до налога, руб.": [-adaptation_loss] + [annual_profit_before_tax] * evaluation_years,
    "Налог, руб.": [0] + [annual_tax] * evaluation_years,
    "Чистый денежный поток, руб.": cash_flows
})

display(cash_flow_df.style.format({col: "{:,.0f}" for col in cash_flow_df.columns if col != "Год"}))

print(f"Чистый денежный поток за каждый год 1–{evaluation_years}: {annual_net_cash_flow:,.0f} руб.")
```

### Интерпретация

Отрицательный поток года 0 отражает потерю производительности в период адаптации. В каждом последующем году положительный эффект уменьшается на стоимость лицензий и налог на прибыль. В этой модели чистый годовой денежный поток одинаков для всех трёх лет, поскольку ежегодный эффект от продаж зафиксирован на уровне 8% от текущего оборота.

## 6. NPV проекта

Чистая приведённая стоимость рассчитывается как сумма дисконтированных денежных потоков с учётом первоначальных затрат в году 0:

\[NPV = CF_0 + \sum_{t=1}^{T}\frac{CF_t}{(1+r)^t}\]

где \(CF_t\) — денежный поток года \(t\), \(r\) — ставка дисконтирования 12%, \(T\) — горизонт оценки 3 года.

```python
discounted_cash_flows = [
    cash_flows[year] / (1 + discount_rate) ** year
    for year in range(evaluation_years + 1)
]
npv = sum(discounted_cash_flows)

npv_df = pd.DataFrame({
    "Год": list(range(evaluation_years + 1)),
    "Денежный поток, руб.": cash_flows,
    "Коэффициент дисконтирования": [1 / (1 + discount_rate) ** year for year in range(evaluation_years + 1)],
    "Дисконтированный поток, руб.": discounted_cash_flows
})
display(npv_df.style.format({"Денежный поток, руб.": "{:,.0f}", "Коэффициент дисконтирования": "{:.4f}", "Дисконтированный поток, руб.": "{:,.0f}"}))

print(f"NPV проекта: {npv:,.0f} руб.")
if npv > 0:
    print("NPV положительный: в рамках принятых допущений проект создаёт стоимость.")
else:
    print("NPV неположительный: в рамках принятых допущений проект не создаёт достаточной стоимости.")
```

## 7. Срок окупаемости

Простой срок окупаемости рассчитывается по недисконтированным потокам, а дисконтированный — по дисконтированным. Если годовой поток условно поступает равномерно в течение года, дробная часть срока определяется отношением невозмещённой суммы к годовому потоку.

```python
simple_payback = adaptation_loss / annual_net_cash_flow
discounted_year1_flow = discounted_cash_flows[1]
discounted_payback = adaptation_loss / discounted_year1_flow

print(f"Простой срок окупаемости: {simple_payback:.3f} года ({simple_payback * 12:.1f} месяца)")
print(f"Дисконтированный срок окупаемости: {discounted_payback:.3f} года ({discounted_payback * 12:.1f} месяца)")

cumulative_cash_flow = []
running_total = 0
for flow in cash_flows:
    running_total += flow
    cumulative_cash_flow.append(running_total)

cumulative_df = pd.DataFrame({"Год": list(range(evaluation_years + 1)), "Накопленный денежный поток, руб.": cumulative_cash_flow})
display(cumulative_df.style.format({"Накопленный денежный поток, руб.": "{:,.0f}"}))
```

### Важное замечание к методике

В некоторых учебных решениях к дробной части срока окупаемости прибавляют один год, хотя первоначальные затраты уже возмещаются внутри первого года. Здесь срок считается как отношение первоначальной потери к годовому чистому потоку, поскольку поток первого года превышает затраты на адаптацию. Дисконтированный срок также рассчитан отдельно.

## 8. ROI проекта

Для сопоставимости используем определение: ROI = чистый финансовый результат проекта за период / общие затраты проекта × 100%. Чистый финансовый результат — сумма чистых годовых денежных потоков за три года за вычетом первоначальной потери на адаптацию. Общие затраты включают адаптацию и все лицензии за три года. Годовая стоимость лицензий уже учтена при расчёте чистых денежных потоков, поэтому повторно вычитать её из результата нельзя.

```python
total_net_cash_inflows = annual_net_cash_flow * evaluation_years
project_net_result = total_net_cash_inflows - adaptation_loss
total_project_costs = adaptation_loss + total_license_cost
roi = project_net_result / total_project_costs * 100

print(f"Сумма чистых потоков за годы 1–{evaluation_years}: {total_net_cash_inflows:,.0f} руб.")
print(f"Чистый финансовый результат после затрат на адаптацию: {project_net_result:,.0f} руб.")
print(f"Общие затраты (адаптация + лицензии): {total_project_costs:,.0f} руб.")
print(f"ROI проекта: {roi:.1f}%")
```

## 10. Анализ нематериальных эффектов

Внедрение CRM может дать следующие нематериальные выгоды:

- **Повышение качества обслуживания:** история взаимодействий и напоминания помогают быстрее отвечать клиентам.
- **Рост лояльности клиентов:** менеджеры могут использовать историю покупок и персонализировать предложения.
- **Снижение числа ошибок:** единая база уменьшает дублирование и потерю информации.
- **Повышение прозрачности процессов:** руководитель видит воронку продаж, загрузку менеджеров и причины отказов.
- **Более точное планирование:** отчёты и аналитика помогают прогнозировать спрос и выручку.
- **Удовлетворённость сотрудников:** автоматизация повторяющихся операций позволяет уделять больше времени клиентам.

Эти эффекты не всегда сразу отражаются в бухгалтерской отчётности, но могут влиять на удержание клиентов, повторные продажи и качество управленческих решений.

## 11. Парадокс производительности Солоу

Парадокс производительности Солоу связан с тем, что инвестиции в информационные технологии не всегда приводят к немедленному и очевидному росту измеряемой производительности.

В случае CRM это может проявляться следующим образом:

1. **Временной лаг:** сотрудники сначала тратят время на обучение и ввод данных, поэтому краткосрочная производительность может снизиться.
2. **Проблемы измерения:** улучшение качества обслуживания и клиентского опыта трудно сразу выразить в рублях.
3. **Неизменённые процессы:** если компания автоматизирует неэффективные процессы, а не пересматривает их, выгода будет ограниченной.
4. **Недостаточное использование системы:** если менеджеры ведут данные вне CRM или заполняют карточки формально, аналитика будет неполной.
5. **Дополнительные организационные меры:** результат зависит от обучения, интеграции с другими системами, регламентов и KPI.

Следовательно, фактический эффект может оказаться ниже расчётного или проявиться позже. Для проверки результата следует сравнивать показатели до и после внедрения: время обработки лида, конверсию по этапам воронки, длительность сделки, повторные продажи и удовлетворённость клиентов.
